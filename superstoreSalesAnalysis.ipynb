<h3> EXPLORATORY DATA ANALYSIS

<h6> WHAT IS THE OVERALL SALES TREND

```python
df['order_date'].min()
```

```python
df['order_date'].max()
```

```python
#getting month year from the data set
df['month_year']=df['order_date'].apply(lambda x :x.strftime('%Y-%m') )
```

```python
#Grouping month_year
df_trend=pd.DataFrame(df.groupby('month_year')['sales'].sum()).reset_index()
```

```python
# plotting sales in each month
plt.figure(figsize=(16,5))
plt.plot(df_trend['month_year'],df_trend['sales'],color='red')
plt.xticks(rotation='vertical')
plt.show()
```

<h5> WHICH ARE TOP 10 PRODUCTS BY SALES

```python
#grouping based on product name
prod_sales=pd.DataFrame(df.groupby('product_name')['sales'].sum())
```

```python
#finding top 10 products by sales
prod_sales.sort_values('sales',ascending=False)[:10]
```

<h5>WHICH ARE THE MOST SELLING PRODUCTS?

```python
#grouping by product name and quantity
prod_quan=pd.DataFrame(df.groupby('product_name')['quantity'].sum())
```

```python
#finding most selling products
prod_quan.sort_values('quantity',ascending=False)[:10]
```

<h5>WHAT IS THE MOST PREFFERED SHIP MODE

```python
plt.figure(figsize=(8,5))
sns.countplot(df['ship_mode'])
plt.show()
```

<h5>WHAT IS THE MOST PROFFITABLE CATOGORY AND SUB CATOGORY

```python
df_cate=pd.DataFrame(df.groupby(['category','sub_category'])['profit'].sum())
```

```python
df_cate.sort_values(['category','profit'],ascending=False)
```
